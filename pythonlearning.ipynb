```python
year_grouped = df.groupby('Outlet Establishment Year')['Total Sales'].sum().reset_index()

sns.set(style='whitegrid')
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=year_grouped,
    x='Outlet Establishment Year',
    y='Total Sales',
    marker='o',
    color='dodgerblue'
)

plt.xlabel('Outlet Year', fontsize=12)
plt.ylabel('Total Sales', fontsize=12)
plt.title('Total Sales by Outlet Establishment Year', fontsize=14)
plt.grid(True)

# Customize x-ticks (e.g., show every 2nd year to reduce clutter)
plt.xticks(year_grouped['Outlet Establishment Year'][::2])

# Add annotations for each data point
for idx, row in year_grouped.iterrows():
    plt.annotate(
        f"{row['Total Sales']:.0f}",  # format as integer
        (row['Outlet Establishment Year'], row['Total Sales']),
        textcoords="offset points",
        xytext=(0, 8),  # vertical offset
        ha='center',
        fontsize=9,
        color='black'
    )

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=year_grouped,
    x='Outlet Establishment Year',
    y='Total Sales',
    palette='bright'
)

plt.title('Total Sales By Year', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Total Sales', fontsize=12)

# Show grid lines only on y-axis for better readability
ax.grid(True, axis='y', linestyle='--', alpha=0.7)

# Add value labels on top of each bar
for p in ax.patches:
    height = p.get_height()
    ax.annotate(
        f'${int(height):,}',  # Format with commas for thousands
        (p.get_x() + p.get_width() / 2, height),
        ha='center',
        va='bottom',
        fontsize=10,
        color='black',
        xytext=(0, 5),
        textcoords='offset points'
    )

plt.tight_layout()
plt.show()
```

```python
# Set the visual style
sns.set(style="whitegrid")

# Create the scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='Rating',
    y='Total Sales',
    hue='Item Fat Content',      # Color points by fat content
    alpha=0.7,                   # Add transparency
    palette='Set2'
)

# Title and axis labels
plt.title('Rating vs Total Sales by Fat Content', fontsize=14)
plt.xlabel('Product Rating')
plt.ylabel('Total Sales ($)')
plt.legend(title='Fat Content', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Create a pivot table
pivot = df.pivot_table(
    index='Outlet Establishment Year',
    columns='Item Fat Content',
    values='Total Sales',
    aggfunc='sum'
)

# Plot the heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(
    pivot,
    annot=True,
    cmap='YlGnBu',
    fmt=".0f"
)
plt.title('Total Sales by Year & Fat Content')
plt.tight_layout()
plt.show()
```
